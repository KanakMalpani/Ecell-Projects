```python
import pandas as pd
from huggingface_hub import hf_hub_download

path = hf_hub_download(
    "winterForestStump/10-K_sec_filings",
    "data/026-00000-of-00001-617edf0addd6f69e.parquet",
    repo_type="dataset",
)
df = pd.read_parquet(path)
print(df.shape)
df.head()
```

```python
# check which text columns are mostly empty
text_cols = ["Business", "Risk Factors", "Financial Statements and Supplementary Data"]
for col in text_cols:
    filled = df[col].notna().sum()
    print(col, filled)
```
