# Exploratory LLM Notebook — Task 3 CRM

**Purpose:** Sandbox for experimenting with Ollama/Gemini prompts, testing ticket summarization, and prototyping LangGraph agent nodes outside the main pipeline.

**How to use:**
1. Start Ollama: `ollama serve` + `ollama pull llama3:8b-instruct`
2. Run `python run_pipeline.py` first to populate the database
3. Import from src: `from src.llm import llm_client`, `from src.agents import agent_workflow`

**PI Interview talking point:** Notebooks are for exploration; production logic lives in `src/agents.py` and `src/llm.py`.
